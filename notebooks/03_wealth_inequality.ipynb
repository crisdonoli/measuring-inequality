# 03 — Wealth Inequality (Classic)

Uses `efh_2021_clean.parquet` to estimate survey-weighted wealth shares (bottom 50%, top 10%, top 1%) without tail correction.

```python
import matplotlib.pyplot as plt
import pandas as pd

from utils import load_root
ROOT = load_root()

from source import shares as wealth
from source import figures as figplots

%load_ext autoreload
%autoreload 2
```

```python
DATAPATH = ROOT / 'data/processed'
FIGURESPATH = ROOT / 'figures/p2/'

efh_df = pd.read_parquet(DATAPATH / 'efh_2021_clean.parquet')
```

**2.i Using the Household Finance Survey (EFH) of the Central Bank of Chile, estimate wealth inequality for households in Chile for the year 2021.**

```python
res = wealth.weighted_shares(efh_df, wealth_col='net_wealth')
```

```python
fig, ax = plt.subplots(1, 1, figsize=(6,1.8))
fig, ax = figplots.plot_shares(res, fig, ax)
figplots.save_fig(FIGURESPATH/'share.pdf', fig, dpi=300)
plt.tight_layout()
plt.show()
```
